```python
print("===== Travel Agent =====")

name = input("Enter your name: ")
destination = input("Enter your preferred destination: ")
budget = float(input("Enter your budget (USD): "))
days = int(input("Enter number of days: "))

print("\n===== Travel Plan =====")
print("Name:", name)
print("Destination:", destination)
print("Budget: $", budget)
print("Number of Days:", days)

# Recommendation
if budget < 500:
    hotel = "Budget Hotel"
    transport = "Public Transport"
elif budget < 1000:
    hotel = "Standard Hotel"
    transport = "Bus / Train"
else:
    hotel = "Luxury Hotel"
    transport = "Flight / Private Transport"

print("\nRecommended Hotel:", hotel)
print("Recommended Transport:", transport)
print("\nHave a safe and enjoyable trip!")
```
